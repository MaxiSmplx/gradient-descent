# Dilemma of Local Minima

Gradient Descent has a fundamental limitation when optimizing non-convex functions: local minima.

When approximating the optimal values for parameters or coefficients $w_i$ of a function $f$, we update the parameters in the direction of steepest descent based on the gradient of the function, given by $\frac{df}{dw}$. The objective is to find the global minimum of the function.

The global minimum yields the optimal parameters or solution to the given problem, e.g. coefficients for Linear Regression with the smallest (minimum) error between the model's predictions and the observed values.

However, because Gradient Descent only follows the local direction of steepest descent, it can converge to a local minimum rather than the global minimum. The result can therefore depend on the initial parameter values.

The figure below illustrates how changing the initial value of $x$ from $1$ to $-0.5$ affects the progression of Gradient Descent.

As a general optimization problem, there is no guarantee that Gradient Descent will find the global minimum when optimizing a non-convex function. Without additional information about the objective function, we cannot determine solely from the gradient whether a minimum is local or global. Consequently, the initialization of the parameters can influence which minimum Gradient Descent converges to.

For Linear Regression and Logistic Regression however, this problem does not persist. Both have convex loss functions that have, by definition, only one minimum: the global minimum.

More complex polynomial function, used in Neural Networks, do have dozens of local minima, to which this dilemma applies.

```python
def f(x):
    """f(x) = 0.1x^4 - 2x^2 + x + 5"""
    return (0.1 * x**4) - (2 * x **2) + x + 5

def df(x):
    """f(X) = 0.4x^3 - 4x + 1"""
    return (0.4 * x**3) - (4 * x) + 1
```
